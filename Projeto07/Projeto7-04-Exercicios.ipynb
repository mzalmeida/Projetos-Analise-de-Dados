## Carregando os Dados

```python
df = pd.read_csv("dados/Processamento/animeData.csv")
```

```python
df.head(750)
```

```python
# Verificar o tipo de dados categoricos
df.dtypes 
```

```python
# 1- Qual a média de score dos animes?

# Calcular a média da coluna 'score'
media_score = df["score"].mean()
print("Média da coluna score:", media_score)
```

```python
# 2- Qual o anime com maior número de votos?
# https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.max.html

maiorValor = df["votersCount"].max()
print("O voto com maior valor é:", maiorValor)
```

```python
# 3- Quais os tipos de animes?
# https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.value_counts.html


TiposAnimes = df["animeType"].value_counts()
print("Os tipos são:" , TiposAnimes)
```

```python
# 4- Qual a média de score por tipo de anime?
# https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.groupby.html#pandas.DataFrame.groupby
# https://pandas.pydata.org/docs/reference/api/pandas.api.typing.DataFrameGroupBy.mean.html

MediaTipo  = df.groupby("animeType")["score"].mean()
print("A media por tipo é", MediaTipo)
```

```python
# 8- Há animes sem produtores?

#SemoProd = df["Producer"].value_counts()
#print("Producer", SemoProd)


# Contar valores em branco (NaN) em uma linha
print("Valores em branco:", df["Producer"].isna().sum())

# Contar valores preenchidos (não nulos) em uma linha
print("Valores preenchidos:", df["Producer"].count())


```

```python
# 9- Quantos vezes cada produtor participou da produção de um anime?

qtdProducer  = df.groupby("Producer").count()   
print("Quantidade de vezes é ", qtdProducer)
```
