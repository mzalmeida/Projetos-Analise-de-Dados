## Aplicando One-Hot-Encoding

https://pandas.pydata.org/docs/reference/api/pandas.get_dummies.html

```python
# Dataset original
dados
```

```python
# Obtém as variáveis dummy
# df_dummies = pd.get_dummies(dados['Modelo'])
# Cria variáveis dummies (One-Hot Encoding) a partir da coluna "Modelo" do DataFrame "dados".
# Cada categoria distinta em "Modelo" vira uma nova coluna binária (0 ou 1).
# Exemplo: se "Modelo" tiver valores ["A", "B", "A", "C"],
# o resultado será um DataFrame com colunas "A", "B", "C",
# onde cada linha indica com 1 ou 0 se pertence àquela categoria.
#
# O resultado é armazenado em "df_dummies", que será um DataFrame
# contendo apenas as colunas dummies referentes à variável "Modelo".
df_dummies = pd.get_dummies(dados['Modelo'])
```

```python
# Variáveis dummy
df_dummies
```

```python
# Concatenamos os dataframes
# df = pd.concat([dados, df_dummies], axis='columns')
# Concatena (junta) o DataFrame original "dados" com o DataFrame "df_dummies".
# O parâmetro axis='columns' indica que a junção será feita pelas colunas (lado a lado).
# Assim, o resultado "df" terá todas as colunas originais de "dados"
# mais as colunas dummies criadas a partir da variável "Modelo".
#
# Exemplo:
# Se "dados" tinha colunas [ID, Modelo] e "df_dummies" tinha colunas [A, B, C],
# então "df" terá colunas [ID, Modelo, A, B, C].

df = pd.concat([dados, df_dummies], axis = 'columns')
```

```python
df
```

```python
# Drop da coluna categórica
# newdf = df.drop(['Modelo'], axis=1)
# Cria um novo DataFrame chamado "newdf" a partir de "df",
# removendo a coluna "Modelo".
#
# O parâmetro axis=1 indica que a remoção será feita em colunas (não em linhas).
# Assim, "newdf" terá todas as colunas originais de "df" exceto "Modelo".
#
# Isso é útil porque já transformamos "Modelo" em variáveis dummies,
# então a coluna original categórica não é mais necessária para análise ou modelagem.
newdf = df.drop(['Modelo'], axis = 1)
```

```python
newdf
```

```python
# Define os valores de X
# x = newdf.drop('Preco_Venda', axis=1)
# Cria um novo DataFrame chamado "x" a partir de "newdf",
# removendo a coluna "Preco_Venda".
#
# O parâmetro axis=1 indica que a remoção será feita em colunas (não em linhas).
# Assim, "x" conterá todas as variáveis independentes (features),
# enquanto "Preco_Venda" será tratada separadamente como variável alvo (y).
#
# Esse passo é típico em projetos de Machine Learning:
# - "x" → conjunto de variáveis explicativas (inputs).
# - "y" → variável que queremos prever (output).

x = newdf.drop('Preco_Venda', axis = 1)
x
```

```python
# Define os valores de y
# y = newdf['Preco_Venda']
# Cria a variável alvo (y) a partir da coluna "Preco_Venda" do DataFrame "newdf".
# Essa coluna contém os valores que queremos prever (preço de venda).
#
# Agora temos:
#   - x → conjunto de variáveis independentes (features), ou seja, os atributos que explicam o preço.
#   - y → variável dependente (target), ou seja, o preço de venda que será previsto pelo modelo.
#
# Esse é o padrão em projetos de Machine Learning:
#   x = dados de entrada
#   y = saída/resultado que queremos estimar

y = newdf['Preco_Venda']
y
```

```python
# Cria o modelo
# modelo_v1 = linear_model.LinearRegression()
# Cria uma instância do modelo de Regressão Linear da biblioteca scikit-learn.
# Esse modelo será usado para aprender a relação entre as variáveis independentes (x)
# e a variável dependente (y = Preco_Venda).
#
# A regressão linear busca ajustar uma linha (ou hiperplano, em múltiplas variáveis)
# que melhor explica os dados, minimizando os erros entre os valores previstos e os reais.
#
# "modelo_v1" agora é um objeto que pode ser treinado (fit) com os dados
# e depois usado para fazer previsões (predict).

modelo_v1 = linear_model.LinearRegression()
```

```python
# Treina o modelo
# Aprendizado supervisionado em machine learning
# modelo_v1.fit(x, y)
# Treina (ajusta) o modelo de regressão linear "modelo_v1"
# usando os dados de entrada (x) e a variável alvo (y).
#
# O método .fit():
#   - Calcula os coeficientes da regressão linear.
#   - Encontra a linha/hiperplano que melhor explica a relação
#     entre as variáveis independentes (x) e o preço de venda (y).
#
# Depois desse passo, o modelo já está treinado e pronto para:
#   - Fazer previsões com .predict(x_novo).
#   - Consultar os coeficientes com .coef_ e o intercepto com .intercept_.

modelo_v1.fit(x,y)
```

```python
# Previsões do modelo
# modelo_v1.predict(x)
# Usa o modelo de regressão linear já treinado (modelo_v1)
# para gerar previsões da variável alvo (Preço de Venda),
# com base nos dados de entrada (x).
#
# O método .predict():
#   - Aplica os coeficientes e o intercepto calculados no treinamento.
#   - Retorna um array com os valores previstos de "Preco_Venda"
#     para cada linha do DataFrame x.
#
# Em resumo: você obtém os preços de venda estimados pelo modelo
# a partir das variáveis explicativas (features) que estavam em x.

modelo_v1.predict(x)
```

```python
# Uma previsão
# modelo_v1.predict([[7000, 4, 0, 0, 1]])
# Faz uma previsão do preço de venda usando o modelo treinado.
# O método .predict() recebe uma lista com os valores das variáveis independentes (features).
#
# Nesse exemplo, os valores passados correspondem a uma única observação:
#   - 7000 → pode ser quilometragem, valor numérico ou outra variável contínua.
#   - 4    → pode ser número de portas, idade do carro, ou outra característica.
#   - 0, 0 → variáveis dummies (categorias que não se aplicam).
#   - 1    → variável dummy indicando a categoria ativa.
#
# O modelo aplica os coeficientes aprendidos no treinamento e retorna
# o valor previsto de "Preco_Venda" para essa combinação de atributos.
modelo_v1.predict([[7000,4,0,0,1]])
```

```python
# Uma previsão
# modelo_v1.predict([[45000, 4, 0, 1, 0]])
# Faz uma previsão do preço de venda para uma nova observação,
# usando o modelo de regressão linear já treinado.
#
# Os valores dentro da lista representam as variáveis independentes (features):
#   - 45000 → pode ser quilometragem, idade ou outra variável contínua.
#   - 4     → pode ser número de portas, anos de uso, etc.
#   - 0, 1, 0 → variáveis dummies que indicam a categoria do "Modelo".
#
# O método .predict() aplica os coeficientes aprendidos e retorna
# o valor estimado de "Preco_Venda" para esse conjunto de características.
modelo_v1.predict([[45000,4,0,1,0]])
```

```python
# Uma previsão
# modelo_v1.predict([[86000, 7, 1, 0, 0]])
# Faz uma previsão do preço de venda para uma nova observação,
# usando o modelo de regressão linear já treinado.
#
# Os valores dentro da lista representam as variáveis independentes (features):
#   - 86000 → valor contínuo (ex.: quilometragem, idade ou outra característica numérica).
#   - 7     → pode ser número de portas, anos de uso ou outra variável inteira.
#   - 1, 0, 0 → variáveis dummies que indicam a categoria do "Modelo".
#
# O método .predict() aplica os coeficientes aprendidos no treinamento
# e retorna o valor estimado de "Preco_Venda" para esse conjunto de atributos.
modelo_v1.predict([[86000,7,1,0,0]])
```

```python
# Acurácia do modelo
# modelo_v1.score(x, y)
# Calcula o coeficiente de determinação R² do modelo de regressão linear.
#
# O método .score():
#   - Compara os valores previstos pelo modelo com os valores reais (y).
#   - Retorna o R², que mede a proporção da variância de "Preco_Venda"
#     explicada pelas variáveis independentes (x).
#
# Interpretação do R²:
#   - 1.0 → modelo perfeito (explica 100% da variância).
#   - 0.0 → modelo não explica nada além da média.
#   - Valores negativos → modelo pior que simplesmente usar a média.
modelo_v1.score(x,y)
```

```python
# Aplica o encoder
# df_car['Modelo'] = le.fit_transform(df_car['Modelo'])
# Aplica o LabelEncoder na coluna "Modelo" do DataFrame "df_car".
# O método .fit_transform():
#   - "fit" → aprende todas as categorias únicas presentes em "Modelo".
#   - "transform" → substitui cada categoria por um número inteiro.
#
# Exemplo:
#   Se "Modelo" tiver valores ["Sedan", "SUV", "Hatch"],
#   o LabelEncoder pode atribuir:
#     Sedan → 0
#     SUV   → 1
#     Hatch → 2
#
# O resultado é que a coluna "Modelo" passa a conter apenas números,
# o que facilita o uso em algoritmos de Machine Learning que não aceitam texto.

df_car['Modelo'] = le.fit_transform(df_car['Modelo'])
```

## Aplicando One-Hot-Encoding com Transformer

https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.OneHotEncoder.html

https://scikit-learn.org/stable/modules/generated/sklearn.compose.ColumnTransformer.html

```python
# Carrega os dados
dados = pd.read_csv('dados/dataset.csv')
dados
```

```python
# from sklearn.preprocessing import OneHotEncoder
#from sklearn.compose import ColumnTransformer

# OneHotEncoder:
#   - Transforma variáveis categóricas em variáveis binárias (dummies).
#   - Exemplo: "Modelo" com categorias ["Sedan", "SUV", "Hatch"]
#     vira colunas extras:
#       Sedan → [1,0,0]
#       SUV   → [0,1,0]
#       Hatch → [0,0,1]

# ColumnTransformer:
#   - Permite aplicar diferentes transformações em diferentes colunas do DataFrame.
#   - Exemplo: aplicar OneHotEncoder na coluna "Modelo"
#     e deixar "Kilometragem" e "Idade_Veiculo" sem transformação.
#
# Isso é útil porque você pode preparar os dados de forma flexível,
# combinando variáveis categóricas (transformadas em dummies)
# com variáveis numéricas (mantidas como estão).

from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
# Cria o encoder com transformação
# OHEncoder = ColumnTransformer(
#    [('Modelo', OneHotEncoder(), [0])],
#    remainder='passthrough'
# )
# Cria um ColumnTransformer chamado "OHEncoder".
#
# O que acontece aqui:
#   - A transformação "Modelo" aplica o OneHotEncoder na coluna de índice 0
#     (ou seja, a primeira coluna da matriz de entrada).
#   - O OneHotEncoder converte categorias em variáveis binárias (dummies).
#   - O parâmetro remainder='passthrough' mantém todas as outras colunas
#     sem transformação (elas passam direto).
#
# Exemplo prático:
#   Se a coluna 0 tiver valores ["Sedan", "SUV", "Hatch"],
#   o OneHotEncoder gera colunas extras:
#     Sedan → [1,0,0]
#     SUV   → [0,1,0]
#     Hatch → [0,0,1]
#
# Enquanto isso, as demais colunas (como Kilometragem e Idade_Veiculo)
# continuam intactas.
OHEncoder = ColumnTransformer([('Modelo', OneHotEncoder(), [0])], remainder = 'passthrough')
```

```python
# Cópia do dataframe original
df_car = dados
```

```python
df_car
```

```python
# Define os valores de X
# x = df_car.drop('Preco_Venda', axis=1)
# Cria a matriz de variáveis independentes (features) removendo a coluna alvo "Preco_Venda".
#
# O método .drop():
#   - Remove a coluna especificada do DataFrame.
#   - O parâmetro axis=1 indica que estamos removendo uma coluna (não uma linha).
#
# Resultado:
#   - "x" contém todas as colunas explicativas (Modelo, Kilometragem, Idade_Veiculo, etc.).
#   - A coluna "Preco_Venda" é excluída porque ela será usada como variável alvo (y).
#
# Esse é o formato padrão para treinar modelos de Machine Learning:
#   - x → dados de entrada (features)
#   - y → saída/resultado esperado (target)

x = df_car.drop('Preco_Venda', axis = 1)
x
```

```python
# Define os valores de y
# y = df_car['Preco_Venda']
# Define a variável alvo (y) como a coluna "Preco_Venda" do DataFrame "df_car".
#
# Essa coluna contém os valores reais de preço de venda,
# que o modelo tentará prever a partir das variáveis independentes (x).
#
# Agora temos:
#   - x → matriz de features (todas as colunas exceto Preco_Venda).
#   - y → vetor alvo (Preço de Venda).
#
# Esse é o formato padrão para treinar modelos de Machine Learning:
#   x = dados de entrada
#   y = saída/resultado esperado
y = df_car['Preco_Venda']
y
```

```python
# Aplica o encoder ao dataset x
# X = OHEncoder.fit_transform(x)
# Aplica o ColumnTransformer "OHEncoder" sobre os dados de entrada "x".
#
# O método .fit_transform():
#   - "fit" → aprende as categorias presentes na coluna indicada (no caso, "Modelo").
#   - "transform" → converte essa coluna em variáveis binárias (dummies).
#
# Resultado:
#   - "X" será uma matriz NumPy (ou sparse matrix) contendo:
#       • As colunas dummies geradas pelo OneHotEncoder para "Modelo".
#       • As demais colunas (Kilometragem, Idade_Veiculo, etc.) preservadas.
#
# Exemplo:
#   Se "Modelo" tiver categorias ["Sedan", "SUV", "Hatch"],
#   o OneHotEncoder gera colunas extras:
#       Sedan → [1,0,0]
#       SUV   → [0,1,0]
#       Hatch → [0,0,1]
#   E junto disso permanecem os valores numéricos das outras variáveis.
X = OHEncoder.fit_transform(x)
X
```

```python
# Cria a versão 3 do modelo
# from sklearn import linear_model

# modelo_v3 = linear_model.LinearRegression()
# Cria uma nova instância de regressão linear chamada "modelo_v3".
#
# Esse objeto é independente dos modelos anteriores (modelo_v1 e modelo_v2).
# Isso significa que você pode treinar "modelo_v3" com os dados já transformados
# pelo OneHotEncoder (matriz X), sem interferir nos outros modelos.
#
# A regressão linear ajustará uma equação que relaciona as variáveis independentes
# (categorias de "Modelo" transformadas em dummies + Kilometragem + Idade_Veiculo)
# com a variável alvo (Preco_Venda).
modelo_v3 = linear_model.LinearRegression()
```

```python
# Treina o modelo
# modelo_v3.fit(X, y)
# Treina (ajusta) o modelo de regressão linear "modelo_v3"
# usando os dados de entrada já transformados (X) e a variável alvo (y).
#
# O método .fit():
#   - Calcula os coeficientes da regressão linear considerando
#     as variáveis categóricas convertidas em dummies (via OneHotEncoder)
#     e as variáveis numéricas originais.
#   - Encontra a equação que melhor explica como as variáveis independentes
#     influenciam o "Preco_Venda".
#
# Depois desse passo, "modelo_v3" está treinado e pronto para:
#   - Fazer previsões com .predict(X).
#   - Consultar os coeficientes com .coef_ e o intercepto com .intercept_.
modelo_v3.fit(X,y)
```

```python
# Uma previsão do modelo
# modelo_v3.predict([[0,0,1,45000,4]])
# Usa o modelo de regressão linear "modelo_v3" já treinado
# para prever o preço de venda de um carro com as seguintes características:
#
#   [0, 0, 1, 45000, 4]
#
# Interpretação dos valores:
#   - [0,0,1] → representação OneHotEncoder da categoria "Modelo"
#               (nesse caso, corresponde ao terceiro tipo de modelo cadastrado).
#   - 45000   → valor da quilometragem do veículo.
#   - 4       → idade do veículo em anos.
#
# O método .predict() aplica os coeficientes aprendidos durante o treinamento
# e retorna o preço estimado (em unidades da variável alvo "Preco_Venda").
modelo_v3.predict([[0,0,1,45000,4]])
```

```python
# Uma previsão do modelo
# modelo_v3.predict([[0,1,0,86000,7]])
# Usa o modelo de regressão linear "modelo_v3" já treinado
# para prever o preço de venda de um carro com as seguintes características:
#
#   [0, 1, 0, 86000, 7]
#
# Interpretação dos valores:
#   - [0,1,0] → representação OneHotEncoder da categoria "Modelo"
#               (nesse caso, corresponde ao segundo tipo de modelo cadastrado).
#   - 86000   → quilometragem do veículo.
#   - 7       → idade do veículo em anos.
#
# O método .predict() aplica os coeficientes aprendidos durante o treinamento
# e retorna o preço estimado (em unidades da variável alvo "Preco_Venda").
modelo_v3.predict([[0,1,0,86000,7]])
```

```python
# Uma previsão do modelo
# modelo_v3.score(X, y)
# Calcula o coeficiente de determinação R² para o modelo "modelo_v3".
#
# O método .score():
#   - Compara os valores previstos pelo modelo com os valores reais (y).
#   - Retorna o R², que mede a proporção da variância de "Preco_Venda"
#     explicada pelas variáveis independentes (X).
#
# Interpretação do R²:
#   - 1.0 → modelo perfeito (explica 100% da variância).
#   - 0.0 → modelo não explica nada além da média.
#   - Valores negativos → modelo pior que simplesmente usar a média.
modelo_v3.score(X,y)
```
