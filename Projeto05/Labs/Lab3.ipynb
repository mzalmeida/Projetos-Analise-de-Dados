```python
# Categorias únicas por coluna
# Muitas categorias: count frequency
# poucas categorias: label encoding
# numero intermediario de categorias: hot encoding

# O código abaixo percorre todas as colunas do DataFrame "dados"
# e imprime o nome da coluna junto com a quantidade de categorias únicas.

# for cols in dados.columns:
#     - Aqui usamos um loop "for" para iterar sobre cada elemento em dados.columns.
#     - dados.columns retorna uma lista com os nomes de todas as colunas do DataFrame.
#     - Assim, a variável "cols" assume o nome de cada coluna a cada iteração.

# print(cols,':', len(dados[cols].unique()), 'categorias')
#     - Para cada coluna "cols", o comando print exibe:
#         1. O nome da coluna (cols).
#         2. Dois pontos ":" como separador visual.
#         3. O número de categorias únicas presentes nessa coluna.
#            - dados[cols] acessa a coluna pelo nome.
#            - dados[cols].unique() retorna um array com os valores distintos dessa coluna.
#            - len(dados[cols].unique()) conta quantos valores únicos existem.
#         4. A palavra "categorias" para indicar que o número corresponde
#            à quantidade de categorias distintas.

# -----------------------------
# Exemplo prático:
# Suponha que o DataFrame "dados" tenha duas colunas:
#   "Cor" = [Azul, Azul, Verde, Vermelho]
#   "Tamanho" = [P, M, P, G]
#
# O loop imprimirá:
#   Cor : 3 categorias
#   Tamanho : 3 categorias
#
# Isso mostra quantas categorias diferentes existem em cada coluna do DataFrame.

for cols in dados.columns:
    print(cols,':', len(dados[cols].unique()), 'categorias')
```

```python
# Vamos descobrir quantas colunas adicionais iremos obter após aplicar One-Hot-Encoding
# 123 (total de categorias) - 6 colunas originais = 117

# O código pd.get_dummies(dados, drop_first = True).shape faz o seguinte:

# 1. pd.get_dummies(dados, drop_first = True)
#    - Cria variáveis dummies (também chamadas de variáveis indicadoras ou One-Hot Encoding)
#      para todas as colunas categóricas do DataFrame "dados".
#    - Cada categoria distinta vira uma nova coluna binária (0 ou 1).
#    - O parâmetro drop_first = True remove a primeira categoria de cada coluna
#      para evitar multicolinearidade (redundância de informação).
#    - Exemplo: Se a coluna "Cor" tem valores ["Azul", "Verde", "Vermelho"],
#      o One-Hot Encoding com drop_first=True gera duas colunas:
#         Cor_Verde, Cor_Vermelho
#      (Azul é a categoria de referência e não aparece como coluna).

# 2. .shape
#    - Retorna uma tupla com (número de linhas, número de colunas) do DataFrame resultante.
#    - Exemplo: (100, 15) significa que o DataFrame tem 100 linhas e 15 colunas.

# -----------------------------
# Interpretação prática:
# Esse comando mostra qual será o tamanho do DataFrame após aplicar One-Hot Encoding
# com remoção da primeira categoria. É útil para verificar quantas novas variáveis
# foram criadas e se o número de colunas aumentou muito.

pd.get_dummies(dados, drop_first = True).shape
```

```python
# Aplicamos One-Hot-Encoding
# O código novos_dados = pd.get_dummies(dados, drop_first = True) faz o seguinte:

# 1. pd.get_dummies(dados, drop_first = True)
#    - Aplica One-Hot Encoding em todas as colunas categóricas do DataFrame "dados".
#    - Cada categoria distinta vira uma nova coluna binária (0 ou 1).
#    - O parâmetro drop_first = True remove a primeira categoria de cada coluna
#      para evitar multicolinearidade (redundância de informação).
#    - Exemplo: Se a coluna "Cor" tem valores ["Azul", "Verde", "Vermelho"],
#      o One-Hot Encoding com drop_first=True gera duas colunas:
#         Cor_Verde, Cor_Vermelho
#      (Azul é a categoria de referência e não aparece como coluna).

# 2. novos_dados = ...
#    - Cria um novo DataFrame chamado "novos_dados" com as variáveis dummies.
#    - Esse DataFrame é totalmente numérico, pronto para ser usado em algoritmos
#      de Machine Learning que não aceitam variáveis categóricas diretamente.

# -----------------------------
# Interpretação prática:
# Esse comando transforma todas as variáveis categóricas em variáveis numéricas
# usando One-Hot Encoding. O resultado é um DataFrame expandido, com mais colunas,
# mas sem categorias textuais. Isso é essencial para treinar modelos de ML.

novos_dados = pd.get_dummies(dados, drop_first = True)
```

> Precisamos realmente aplicar One-Hot-Encoding em todas as categorias?

```python
# X1 tem 27 categorias. Vamos listar as 10 categorias com mais registros

# O código dados.X1.value_counts(ascending = False).head(10) faz o seguinte:

# 1. dados.X1
#    - Acessa a coluna "X1" do DataFrame "dados".

# 2. .value_counts(ascending = False)
#    - Conta quantas vezes cada valor aparece na coluna "X1".
#    - Retorna uma Series ordenada pela frequência.
#    - O parâmetro ascending = False garante que os valores mais frequentes
#      apareçam primeiro (ordem decrescente).

# 3. .head(10)
#    - Seleciona apenas os 10 primeiros resultados da Series.
#    - Ou seja, mostra as 10 categorias mais frequentes da coluna "X1".

# -----------------------------
# Exemplo prático:
# Se X1 = [A, B, A, C, B, A, D, D, D, E]
# então dados.X1.value_counts(ascending=False) =
#     A    3
#     D    3
#     B    2
#     C    1
#     E    1
#
# Aplicando .head(10), o resultado será:
#     A    3
#     D    3
#     B    2
#     C    1
#     E    1

# -----------------------------
# Interpretação prática:
# Esse comando retorna as 10 categorias mais comuns da coluna "X1",
# junto com suas respectivas contagens. É útil para identificar
# quais valores aparecem com maior frequência.
dados.X1.value_counts(ascending = False).head(10)
```

```python
# Colocamos as 10 categorias em uma nova variável Python
# O código top_10 = [x for x in dados.X1.value_counts().head(10).index] faz o seguinte:

# 1. dados.X1.value_counts()
#    - Conta quantas vezes cada valor aparece na coluna "X1".
#    - Retorna uma Series ordenada por frequência (do maior para o menor por padrão).

# 2. .head(10)
#    - Seleciona apenas os 10 primeiros valores dessa Series.
#    - Ou seja, pega as 10 categorias mais frequentes da coluna "X1".

# 3. .index
#    - Retorna os índices da Series, que correspondem às categorias (valores únicos da coluna).

# 4. [x for x in ...]
#    - É uma list comprehension.
#    - Constrói uma lista contendo cada elemento do índice (ou seja, cada categoria).

# 5. top_10 = ...
#    - Armazena essa lista na variável "top_10".
#    - Resultado: uma lista com as 10 categorias mais frequentes da coluna "X1".

# -----------------------------
# Exemplo prático:
# Se X1 = [A, B, A, C, B, A, D, D, D, E]
# então dados.X1.value_counts().head(10).index = ["A", "D", "B", "C", "E"]
# e top_10 = ["A", "D", "B", "C", "E"]

# -----------------------------
# Interpretação prática:
# Esse comando gera uma lista com as 10 categorias mais comuns da coluna "X1".
# É útil para análises exploratórias, como focar apenas nas categorias mais relevantes
# ou criar gráficos com as principais ocorrências.
top_10 = [x for x in dados.X1.value_counts().head(10).index]
```

```python
top_10
```

```python
# Criamos uma função para criar as novas variáveis com One-Hot-Encoding
# A função one_hot_top10(df, column, top_10) faz o seguinte:

# def one_hot_top10(df, column, top_10):
#    for i in top_10:
#        df[column + "_" + i] = np.where(dados[column] == i, 1, 0)

# 1. Parâmetros:
#    - df: o DataFrame que será modificado.
#    - column: o nome da coluna categórica que será transformada.
#    - top_10: lista com as 10 categorias mais frequentes dessa coluna.

# 2. Loop for i in top_10:
#    - Itera sobre cada categoria presente na lista top_10.

# 3. df[column + "_" + i] = ...
#    - Cria uma nova coluna no DataFrame para cada categoria de top_10.
#    - O nome da nova coluna será "nome_da_coluna_categoria".
#    - Exemplo: se column="X1" e i="A", a nova coluna será "X1_A".

# 4. np.where(dados[column] == i, 1, 0)
#    - Usa NumPy para criar uma coluna binária (dummy).
#    - Se o valor da linha em "column" for igual à categoria i → 1.
#    - Caso contrário → 0.
#    - Exemplo: Se column="X1" e i="A":
#        X1 = [A, B, A, C] → X1_A = [1, 0, 1, 0]

# -----------------------------
# Interpretação prática:
# Essa função aplica um **One-Hot Encoding personalizado** apenas para as 10 categorias
# mais frequentes de uma coluna. Assim, evita criar colunas para categorias raras,
# reduzindo dimensionalidade e mantendo apenas as mais relevantes.

def one_hot_top10(df, column, top_10):
    for i in top_10:
        df[column + "_" + i ] = np.where(dados[column]==i,1,0)
```

```python
# Executamos a função e aplicamos One-Hot-Encoding
one_hot_top10(dados, 'X1', top_10)
```
