## Laboratório Prático - Count/Frequency Encoding

Usamos este dataset como referência e fizemos alterações para o Lab:

https://www.kaggle.com/c/mercedes-benz-greener-manufacturing

```python
# Count/Frequency Encoding é uma técnica de pré-processamento de variáveis categóricas.
# Ela transforma categorias (ex: "azul", "vermelho", "verde") em valores numéricos
# baseados na quantidade ou frequência relativa de ocorrência no dataset.

# -----------------------------
# Diferença entre Count e Frequency Encoding
# -----------------------------
# Count Encoding:
#   - Substitui cada categoria pelo número de vezes que aparece no conjunto de dados.
#   - Exemplo: Se a coluna "Cor" tem Azul (10 vezes), Vermelho (5 vezes), Verde (2 vezes)
#              então Azul = 10, Vermelho = 5, Verde = 2.
#
# Frequency Encoding:
#   - Substitui cada categoria pela proporção relativa (frequência) em relação ao total.
#   - Exemplo: No mesmo caso, total = 17 registros.
#              Azul = 10/17 ≈ 0.59, Vermelho = 5/17 ≈ 0.29, Verde = 2/17 ≈ 0.12.

# -----------------------------
# Vantagens
# -----------------------------
# - Simples de implementar.
# - Preserva informação sobre a importância relativa das categorias.
# - Evita explosão de dimensionalidade (diferente do One-Hot Encoding).
# - Funciona bem em algoritmos baseados em árvores (Random Forest, XGBoost).

# -----------------------------
# Limitações
# -----------------------------
# - Pode introduzir viés se categorias raras forem pouco representadas.
# - Em modelos lineares, pode ser interpretado como uma relação ordinal (o que nem sempre faz sentido).
# - Precisa ser calculado apenas no conjunto de treino para evitar data leakage
#   (vazar informação do teste para o treino).

# -----------------------------
# Interpretação prática
# -----------------------------
# Imagine uma base de clientes com a coluna "Cidade".
# - Se "São Paulo" aparece em 40% dos registros e "Birigui" em 2%,
#   o modelo entende que "São Paulo" é uma categoria mais relevante.
# - Isso ajuda a capturar padrões de concentração de clientes sem precisar criar
#   centenas de colunas extras como no One-Hot Encoding.
```

```python
# Imports
import pandas as pd
import numpy as np
```

```python
# Mostrar todas as colunas do dataframe
pd.set_option('display.max_columns', None)
```

```python
# Carrega os dados
df = pd.read_csv('dados/dataset.csv', usecols = ['X1','X2','X3'])
```

```python
# Shape
df.shape
```

```python
# Amostra dos dados
df.head()
```

```python
# O código len(df.X1.unique()) faz o seguinte:
#
# 1. df.X1
#    - Aqui estamos acessando a coluna chamada "X1" dentro do DataFrame "df".
#    - Supondo que df seja um DataFrame do pandas, df.X1 retorna uma Series
#      contendo todos os valores dessa coluna.
#
# 2. df.X1.unique()
#    - O método .unique() retorna um array com todos os valores distintos
#      (sem repetição) presentes na coluna "X1".
#    - Exemplo: se X1 = [A, B, A, C, B], então df.X1.unique() = [A, B, C].
#
# 3. len(df.X1.unique())
#    - A função len() calcula o tamanho do array retornado por .unique().
#    - Ou seja, conta quantos valores únicos existem na coluna "X1".
#    - Exemplo: no caso anterior, len([A, B, C]) = 3.
#
# -----------------------------
# Interpretação prática:
# Esse comando responde à pergunta:
# "Quantas categorias diferentes existem na coluna X1 do meu DataFrame?"
# É útil para entender a diversidade de valores em uma variável categórica.

len(df.X1.unique())
```

```python
# O código abaixo percorre todas as colunas do DataFrame "df"
# e imprime o nome da coluna junto com a quantidade de categorias únicas.

# for i in df.columns:
#     - Aqui usamos um loop "for" para iterar sobre cada elemento em df.columns.
#     - df.columns retorna uma lista com os nomes de todas as colunas do DataFrame.
#     - Assim, a variável "i" assume o nome de cada coluna a cada iteração.

# print(i,"-", len(df[i].unique()), "categorias")
#     - Para cada coluna "i", o comando print exibe:
#         1. O nome da coluna (i).
#         2. Um traço "-" apenas como separador visual.
#         3. O número de categorias únicas presentes nessa coluna.
#            - df[i] acessa a coluna pelo nome.
#            - df[i].unique() retorna um array com os valores distintos dessa coluna.
#            - len(df[i].unique()) conta quantos valores únicos existem.
#         4. A palavra "categorias" para indicar que o número corresponde
#            à quantidade de categorias distintas.

# -----------------------------
# Exemplo prático:
# Suponha que df tenha duas colunas:
#   "Cor" = [Azul, Azul, Verde, Vermelho]
#   "Tamanho" = [P, M, P, G]
#
# O loop imprimirá:
#   Cor - 3 categorias
#   Tamanho - 3 categorias
#
# Isso mostra quantas categorias diferentes existem em cada coluna do DataFrame.

for i in df.columns:
    print(i,"-", len(df[i].unique()), "categorias")
```

```python
# O código frequencia = df.X1.value_counts().to_dict() faz o seguinte:

# 1. df.X1
#    - Acessa a coluna "X1" do DataFrame "df".
#    - Retorna uma Series com todos os valores dessa coluna.

# 2. df.X1.value_counts()
#    - Conta quantas vezes cada valor aparece na coluna "X1".
#    - Retorna uma Series onde:
#        - O índice são os valores únicos da coluna.
#        - Os valores são as contagens de cada categoria.
#    - Exemplo: Se X1 = [A, B, A, C, B, A]
#               então df.X1.value_counts() = 
#                   A    3
#                   B    2
#                   C    1

# 3. .to_dict()
#    - Converte essa Series em um dicionário Python.
#    - As chaves do dicionário serão os valores únicos da coluna.
#    - Os valores do dicionário serão as contagens correspondentes.
#    - Exemplo: {"A": 3, "B": 2, "C": 1}

# 4. frequencia = ...
#    - Armazena esse dicionário na variável "frequencia".
#    - Agora você pode acessar diretamente a frequência de cada categoria.
#    - Exemplo: frequencia["A"] retorna 3.

# -----------------------------
# Interpretação prática:
# Esse comando cria um dicionário que mostra quantas vezes cada categoria
# aparece na coluna "X1". É útil para análises de distribuição de dados,
# como verificar quais categorias são mais comuns ou raras.

frequencia = df.X1.value_counts().to_dict()
```

```python
# Visualiza
frequencia
```

```python
# O código df.X1 = df.X1.map(frequencia) faz o seguinte:

# 1. frequencia
#    - É um dicionário criado anteriormente com df.X1.value_counts().to_dict().
#    - Contém cada categoria da coluna "X1" como chave e sua contagem como valor.
#    - Exemplo: {"A": 3, "B": 2, "C": 1}

# 2. df.X1.map(frequencia)
#    - O método .map() aplica uma função ou mapeamento a cada elemento da Series.
#    - Nesse caso, ele substitui cada valor da coluna "X1" pelo valor correspondente
#      no dicionário "frequencia".
#    - Exemplo: Se X1 = [A, B, A, C]
#               e frequencia = {"A": 3, "B": 2, "C": 1}
#               então df.X1.map(frequencia) = [3, 2, 3, 1]

# 3. df.X1 = ...
#    - Atualiza a coluna "X1" do DataFrame com os novos valores mapeados.
#    - Ou seja, a coluna deixa de ter categorias originais e passa a ter
#      os números de frequência de cada categoria.

# -----------------------------
# Interpretação prática:
# Esse comando aplica o Count Encoding na coluna "X1".
# Cada categoria é substituída pelo número de vezes que aparece no dataset.
# Isso transforma variáveis categóricas em valores numéricos,
# permitindo que algoritmos de Machine Learning processem os dados.

df.X1 = df.X1.map(frequencia)
```

```python
df
```

> Fazemos o mesmo para as demais colunas.

```python
# Contagem/frequencia das categorias da coluna X2

# O código frequencia = df.X2.value_counts().to_dict() faz o seguinte:

# 1. df.X2
#    - Acessa a coluna "X2" do DataFrame "df".
#    - Retorna uma Series com todos os valores dessa coluna.

# 2. df.X2.value_counts()
#    - Conta quantas vezes cada valor aparece na coluna "X2".
#    - Retorna uma Series onde:
#        - O índice são os valores únicos da coluna.
#        - Os valores são as contagens de cada categoria.
#    - Exemplo: Se X2 = [A, B, A, C, B, A]
#               então df.X2.value_counts() =
#                   A    3
#                   B    2
#                   C    1

# 3. .to_dict()
#    - Converte essa Series em um dicionário Python.
#    - As chaves do dicionário serão os valores únicos da coluna.
#    - Os valores do dicionário serão as contagens correspondentes.
#    - Exemplo: {"A": 3, "B": 2, "C": 1}

# 4. frequencia = ...
#    - Armazena esse dicionário na variável "frequencia".
#    - Agora você pode acessar diretamente a frequência de cada categoria.
#    - Exemplo: frequencia["A"] retorna 3.

# -----------------------------
# Interpretação prática:
# Esse comando cria um dicionário que mostra quantas vezes cada categoria
# aparece na coluna "X2". É útil para análises de distribuição de dados,
# como verificar quais categorias são mais comuns ou raras.


frequencia = df.X2.value_counts().to_dict()
```

```python
# Agora fazemos o replace de cada categoria pela contagem/frequencia na coluna X2

# O código df.X2 = df.X2.map(frequencia) faz o seguinte:

# 1. frequencia
#    - É o dicionário criado anteriormente com df.X2.value_counts().to_dict().
#    - Contém cada categoria da coluna "X2" como chave e sua contagem como valor.
#    - Exemplo: {"A": 3, "B": 2, "C": 1}

# 2. df.X2.map(frequencia)
#    - O método .map() aplica um mapeamento a cada elemento da Series.
#    - Nesse caso, substitui cada valor da coluna "X2" pelo valor correspondente
#      no dicionário "frequencia".
#    - Exemplo: Se X2 = [A, B, A, C]
#               e frequencia = {"A": 3, "B": 2, "C": 1}
#               então df.X2.map(frequencia) = [3, 2, 3, 1]

# 3. df.X2 = ...
#    - Atualiza a coluna "X2" do DataFrame com os novos valores mapeados.
#    - Ou seja, a coluna deixa de ter categorias originais e passa a ter
#      os números de frequência de cada categoria.

# -----------------------------
# Interpretação prática:
# Esse comando aplica o Count Encoding na coluna "X2".
# Cada categoria é substituída pelo número de vezes que aparece no dataset.
# Isso transforma variáveis categóricas em valores numéricos,
# permitindo que algoritmos de Machine Learning processem os dados.
df.X2 = df.X2.map(frequencia)
```

```python
# Contagem/frequencia das categorias da coluna X3

# O código frequencia = df.X3.value_counts().to_dict() faz o seguinte:

# 1. df.X3
#    - Acessa a coluna "X3" do DataFrame "df".
#    - Retorna uma Series com todos os valores dessa coluna.

# 2. df.X3.value_counts()
#    - Conta quantas vezes cada valor aparece na coluna "X3".
#    - Retorna uma Series onde:
#        - O índice são os valores únicos da coluna.
#        - Os valores são as contagens de cada categoria.
#    - Exemplo: Se X3 = [A, B, A, C, B, A]
#               então df.X3.value_counts() =
#                   A    3
#                   B    2
#                   C    1

# 3. .to_dict()
#    - Converte essa Series em um dicionário Python.
#    - As chaves do dicionário serão os valores únicos da coluna.
#    - Os valores do dicionário serão as contagens correspondentes.
#    - Exemplo: {"A": 3, "B": 2, "C": 1}

# 4. frequencia = ...
#    - Armazena esse dicionário na variável "frequencia".
#    - Agora você pode acessar diretamente a frequência de cada categoria.
#    - Exemplo: frequencia["A"] retorna 3.

# -----------------------------
# Interpretação prática:
# Esse comando cria um dicionário que mostra quantas vezes cada categoria
# aparece na coluna "X3". É útil para análises de distribuição de dados,
# como verificar quais categorias são mais comuns ou raras.
frequencia = df.X3.value_counts().to_dict()
```

```python
# Agora fazemos o replace de cada categoria pela contagem/frequencia na coluna X3

# O código df.X3 = df.X3.map(frequencia) faz o seguinte:

# 1. frequencia
#    - É o dicionário criado anteriormente com df.X3.value_counts().to_dict().
#    - Contém cada categoria da coluna "X3" como chave e sua contagem como valor.
#    - Exemplo: {"A": 3, "B": 2, "C": 1}

# 2. df.X3.map(frequencia)
#    - O método .map() aplica um mapeamento a cada elemento da Series.
#    - Nesse caso, substitui cada valor da coluna "X3" pelo valor correspondente
#      no dicionário "frequencia".
#    - Exemplo: Se X3 = [A, B, A, C]
#               e frequencia = {"A": 3, "B": 2, "C": 1}
#               então df.X3.map(frequencia) = [3, 2, 3, 1]

# 3. df.X3 = ...
#    - Atualiza a coluna "X3" do DataFrame com os novos valores mapeados.
#    - Ou seja, a coluna deixa de ter categorias originais e passa a ter
#      os números de frequência de cada categoria.

# -----------------------------
# Interpretação prática:
# Esse comando aplica o Count Encoding na coluna "X3".
# Cada categoria é substituída pelo número de vezes que aparece no dataset.
# Isso transforma variáveis categóricas em valores numéricos,
# permitindo que algoritmos de Machine Learning processem os dados.
df.X3 = df.X3.map(frequencia)
```

```python
# Verificamos se algum valor nulo foi gerado

# O código df.isnull().any() faz o seguinte:

# 1. df.isnull()
#    - Aplica o método .isnull() ao DataFrame "df".
#    - Retorna um DataFrame booleano do mesmo tamanho, onde:
#        - True indica que a célula está vazia (valor nulo/NaN).
#        - False indica que a célula tem algum valor válido.
#    - Exemplo: Se uma célula da coluna "X1" estiver vazia, será marcada como True.

# 2. .any()
#    - O método .any() é aplicado ao resultado de df.isnull().
#    - Por padrão, ele é aplicado ao longo das colunas (axis=0).
#    - Isso significa que para cada coluna, verifica se existe pelo menos um valor True.
#    - Se houver algum valor nulo na coluna, retorna True para essa coluna.
#    - Caso contrário, retorna False.

# 3. Resultado final
#    - O resultado é uma Series onde:
#        - O índice são os nomes das colunas.
#        - Os valores são True ou False indicando se a coluna possui valores nulos.
#    - Exemplo:
#        X1    True
#        X2    False
#        X3    True
#      Isso significa que X1 e X3 têm valores nulos, enquanto X2 não.

# -----------------------------
# Interpretação prática:
# Esse comando é usado para verificar rapidamente quais colunas do DataFrame
# possuem valores faltantes (NaN). É útil para identificar a necessidade de
# tratamento de dados, como preenchimento ou remoção de valores nulos.
df.isnull().any()
```
